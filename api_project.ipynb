```python
# Creating a function to retrieve data from the API endpoint
def retrieve_youtube_data():

    # All API parameters

    # The API endpoint
    url = "https://youtube138.p.rapidapi.com/channel/videos/"

    # Optional: This is parameter that may be needed to get specific data with the endpoint. 
    querystring = {"id":"UCJ5v_MCY6GNUBTO8-D3XoAg","filter":"videos_latest","hl":"en","gl":"US"}

    # The API headers: This carries the API key and the server hosting the API data.
    # *** YOUR API KEY WILL BE DIFFERENT FROM MINE. USE YOURS FROM RapidAPI to follow along ***
    headers = {
	"x-rapidapi-key": "950922b3b4mshf8fa19a1a80c9bep121817jsne868362a71bc",
	"x-rapidapi-host": "youtube138.p.rapidapi.com"
    }

    # Retrieving data from our API: Send a GET request to the API endpoint
    # Here, we use the "requests" library we imported from the start above.
    response = requests.get(url, headers=headers, params=querystring)

    # Show the response from the API call. This will show just the status_code.
    print(response)

    # Check if the request was successful
    if response.status_code == 200:
        print("API call was successful, data retrieved.")

        # Parse the response content as JSON
        json_data = response.json()

        # The actual data from the retrieved data is in the 'contents'. List of JSON data.
        actual_data = json_data['contents']

        # Using pandas "json_normalize" function to Flatten the JSON data
        df = pd.json_normalize(actual_data) # Parsing the LIST of JSON to a DataFrame
        
        # Save the json normalized data (flat file) 
        # This will be saved as a csv file in your current project folder
        df.to_csv("raw_api_data.csv", index=False)
    else:
        # Here, you can do all possible actions as the response status_code was not "200"
        print("API call was UNSUCCESSFUL. There was an error, please investigate the issue.")

    
    
```

```python
# Calling the function "def retrieve_youtube_data()" we created above
retrieve_youtube_data()
```

## Data Transformation

```python
# Copying our raw data to a staging dataframe.
# Remember as Data Engineers, alawys leave your raw data as it is.
df_contents_staging = df_contents.copy()
```

```python
# Sneak peaking into our staged dataframe (the copy) by looking at just the first two rows
df_contents_staging.head(2)
```

```python
# Copying selected columns into a new dataframe
# Columns of interest to us (based on potential business questions from stakeholders)

# Sample format from ChatGPT: new_df = df[["col1", "col2", "col3"]].copy()

df_contents_staging_new = df_contents_staging[["video.isLiveNow", 
                                               "video.lengthSeconds",
                                               "video.publishedTimeText",
                                               "video.stats.views",
                                               "video.title",
                                               "video.videoId"]].copy()
```

```python
# This new dataframe name "df_contents_staging_new" is too long. 
# Lets shorten it for easy subsequent use.

staged_df = df_contents_staging_new
```

```python
# Sneak peaking into our NEW dataframe by looking at just the first two rows
staged_df.head(2)
```

```python
# Moving through each column one-by-one to see what we can transform as Data Engineers
# This is to help the Data Analysts to answer possible/potential business questions

# Checking the first column in our "df_contents_staging_new" dataframe
# >>> "video.isLiveNow" column

# This checks for all unique data in that column
staged_df['video.isLiveNow'].unique()
```

```python
# This checks for, and counts any null values in that column
staged_df['video.isLiveNow'].isnull().value_counts()

# We can check for more issues/inconsistencies but so far, we think the column is good to go.
```

```python
# Checking the next column in our "df_contents_staging_new" dataframe
# >>> "video.lengthSeconds" column

# After checking, decided to add a new column "vid_length_mins" to the dataset
# This new column will be the "video.lengthSeconds" transformed in minutes
# This might be helpful to our Data Analytics team.
staged_df["vid_length_mins"] = staged_df["video.lengthSeconds"] / 60
```

```python
# Sneak peaking into our NEW dataframe AGAIN by looking at just the first two rows
df_contents_staging_new.head(2)
```

```python
# Our added column "vid_length_mins" is in 6 decimal places. Lets round it up.

# Sample format from ChatGPT: df["col_rounded"] = df["col"].round(2)
staged_df["vid_length_mins"] = (staged_df["video.lengthSeconds"] / 60).round(0)
```
